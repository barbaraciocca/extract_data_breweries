```python
# Install required packages
pip install --upgrade pandas fastparquet s3fs
```

```python
# Insert the state you want to validate here:
state = "Alabama"
```

```python
from datetime import datetime
import pandas as pd

layer = "silver"
today_str = datetime.today().strftime("%Y-%m-%d")

path = f"s3://{layer}/state={state}/breweries_breweries_{today_str}.parquet"

df = pd.read_parquet(
    path,
    engine="fastparquet",
    storage_options={
        "key": "minioadmin",
        "secret": "minioadmin",
        "client_kwargs": {"endpoint_url": "http://localhost:9000/"},
    },
)

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
df
```
