```python
# Set the layer and file name for the gold table
layer = 'gold'
file_name = 'aggregated_breweries.parquet'
```

```python
import pandas as pd

df = pd.read_parquet(
    f"s3://{layer}/{file_name}",
    storage_options={
        "key": "minioadmin",
        "secret": "minioadmin",
        "client_kwargs": {"endpoint_url": "http://localhost:9000/"},
    },
)
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)

df
```
